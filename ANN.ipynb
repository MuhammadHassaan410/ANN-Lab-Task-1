```python
import os
import joblib
import numpy as np
import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (mean_absolute_error,mean_squared_error,r2_score)

print("All libraries imported successfully!")
```

```python
df = pd.read_csv(r"C:\Users\92335\Downloads\auto-mpg.csv")

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

display(df.head())
```

```python
print("Column names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate rows:", df.duplicated().sum())

display(df.describe())
```

```python
df.columns = df.columns.str.strip()

df = df.replace("?", np.nan)

df["horsepower"] = pd.to_numeric(df["horsepower"], errors="coerce")

numeric_columns = ["mpg", "cylinders", "displacement","horsepower", "weight", "acceleration", "model year"]

for column in numeric_columns:
    df[column] = pd.to_numeric(df[column], errors="coerce")

df = df.drop_duplicates()
df = df.dropna(subset=numeric_columns)

print("Dataset cleaned successfully!")
print("Remaining rows:", len(df))

display(df.head())
```

```python
features = ["cylinders","displacement","horsepower","weight","acceleration","model year"]

X = df[features]
y = df["mpg"]

print("Input features:")
print(features)

print("\nTarget variable: mpg")
print("Input shape:", X.shape)
print("Target shape:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.20,random_state=42)

print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
model = LinearRegression()

model.fit(X_train, y_train)

print("Auto MPG Linear Regression model trained successfully!")
```

```python
y_pred = model.predict(X_test)

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)

print("Model Evaluation Results")

print("\nMAE:", round(mae, 3))
print("MSE:", round(mse, 3))
print("RMSE:", round(rmse, 3))
print("R2 Score:", round(r2, 3))
```

```python
coefficients = pd.DataFrame({"Feature": features,"Coefficient": model.coef_})

display(coefficients)

print("Intercept:", round(model.intercept_, 3))
```

```python
sample_car = pd.DataFrame([{"cylinders": 4,"displacement": 140,"horsepower": 90,"weight": 2500,"acceleration": 15,"model year": 76}])

predicted_mpg = model.predict(sample_car)[0]

print("Sample Car Details:")
display(sample_car)

print("Predicted fuel efficiency:", round(predicted_mpg, 2), "MPG")
```

```python
model_data = {"model": model,"features": features}

joblib.dump(model_data, "auto_mpg_model.pkl")

print("Auto MPG model saved successfully!")
```

```python
import gradio as gr

def predict_mpg(
    cylinders,
    displacement,
    horsepower,
    weight,
    acceleration,
    model_year
):
    input_data = pd.DataFrame([{
        "cylinders": cylinders,
        "displacement": displacement,
        "horsepower": horsepower,
        "weight": weight,
        "acceleration": acceleration,
        "model year": model_year
    }])

    prediction = model.predict(input_data)[0]

    return f"Predicted Fuel Efficiency: {prediction:.2f} MPG"

gradio_app = gr.Interface(
    fn=predict_mpg,
    inputs=[
        gr.Number(label="Cylinders", value=4),
        gr.Number(label="Displacement", value=140),
        gr.Number(label="Horsepower", value=90),
        gr.Number(label="Weight", value=2500),
        gr.Number(label="Acceleration", value=15),
        gr.Number(label="Model Year (70–82)", value=76)
    ],
    outputs=gr.Textbox(label="Prediction"),
    title="Auto MPG Prediction",
    description="Predict a car's fuel efficiency using Linear Regression."
)

gradio_app.launch()
```

```python
from flask import Flask, render_template, request

# Load the Auto MPG model
auto_data = joblib.load("auto_mpg_model.pkl")

auto_model = auto_data["model"]
auto_features = auto_data["features"]

# Use a unique Flask app and a separate template folder
auto_app = Flask(
    "AutoMPGApp",
    template_folder="templates_auto"
)

@auto_app.route("/", methods=["GET", "POST"])
def auto_home():
    prediction = None
    error = None

    if request.method == "POST":
        try:
            values = [
                float(request.form[feature])
                for feature in auto_features
            ]

            input_df = pd.DataFrame(
                [values],
                columns=auto_features
            )

            result = auto_model.predict(input_df)[0]
            prediction = round(float(result), 2)

        except (ValueError, KeyError):
            error = "Please enter valid values in all fields."

    return render_template(
        "index.html",
        prediction=prediction,
        error=error
    )

print("Auto MPG Flask application is ready!")
```

```python
auto_app.run(host="127.0.0.1",port=5001,debug=False,use_reloader=False)
```
